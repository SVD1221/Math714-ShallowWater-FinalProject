```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.animation as animation
from matplotlib.colors import ListedColormap
```

```python
from matplotlib import rc
rc('animation', html='jshtml')
colors = ["blue", "red", "white", "pink", "green", "black"]
cmap = ListedColormap(colors)
```

```python
iterations = 1000
N = 20
domain = np.linspace(0, 1, N)
```

```python
fig, ax = plt.subplots()
ax.set_xticks([])
ax.set_yticks([])
ax.set_ylim([0, 2.25])

ims = []
for i in range(iterations+1):
    file = open(f"output/lax_friedrich{i}.txt")
    fileString = file.read().split("\n")
    file.close()    

    height = np.array([float(x) for x in fileString[0].split()])
    
    frame_artists = []
    artist1, = ax.plot(domain, height, color = "blue")
    artist2 = ax.fill_between(domain, height, color= "b", alpha= 0.2)

    frame_artists.append(artist1)
    frame_artists.append(artist2)

    ims.append(frame_artists)
    
ani = animation.ArtistAnimation(fig, ims, blit=True)
ani.save("animations/shallow_water_animation.mp4", writer="ffmpeg", fps=100)
```
