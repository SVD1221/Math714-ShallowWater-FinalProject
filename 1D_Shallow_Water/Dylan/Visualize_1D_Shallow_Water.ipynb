```python
iterations = 1000
N = 10
domain = np.linspace(0, 1, N)
```

```python
def b(x):
    return 0.25*np.sin(np.pi*x)

B = b(domain)
```

```python
fig, ax = plt.subplots()
ax.set_xticks([])
ax.set_yticks([])
ax.set_ylim([0, 2.25])

ims = []
for i in range(iterations):
    file = open(f"output/bathymetry/lax_friedrich{i}.txt")
    fileString = file.read().split("\n")
    file.close()

    height = np.array([float(x) for x in fileString[0].split()])
    
    frame_artists = []
    artist1, = ax.plot(domain, height, color = "blue")
    artist2 = ax.fill_between(domain, height, color= "b", alpha= 0.2)
    artist3, = ax.plot(domain, B, color = "blue")
    artist4 = ax.fill_between(domain, B, color= "r", alpha= 1)

    frame_artists.append(artist1)
    frame_artists.append(artist2)
    frame_artists.append(artist3)
    frame_artists.append(artist4)

    ims.append(frame_artists)
    
ani = animation.ArtistAnimation(fig, ims, blit=True)
ani.save("animations/shallow_water_animation_UNNAMED.mp4", writer="ffmpeg", fps=100)
```
